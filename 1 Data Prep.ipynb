```python
# playlist of top songs from 10 years 2013 to 2022
# there are some songs here which repeat across year - do we include multiple entries?

playlist_urls = {
    2013: 'https://open.spotify.com/playlist/37i9dQZF1DX3Sp0P28SIer', #100 songs
    2014: 'https://open.spotify.com/playlist/37i9dQZF1DX0h0QnLkMBl4', #100 songs
    2015: 'https://open.spotify.com/playlist/37i9dQZF1DX9ukdrXQLJGZ?si=17aa4b3e603d426b', #100 songs
    2016: 'https://open.spotify.com/playlist/37i9dQZF1DX8XZ6AUo9R4R?si=93d023c9de3f42f3', #130 songs
    2017: 'https://open.spotify.com/playlist/37i9dQZF1DWTE7dVUebpUW?si=0b7d5eb26427433a', #100 songs
    2018: 'https://open.spotify.com/playlist/37i9dQZF1DXe2bobNYDtW8?si=6c0a3c1635174dd6', #100 songs
    2019: 'https://open.spotify.com/playlist/37i9dQZF1DWVRSukIED0e9?si=c06f704b3efd4839', #100 songs
    2020: 'https://open.spotify.com/playlist/2fmTTbBkXi8pewbUvG3CeZ', #135 songs (unofficial playlist)
    2021: 'https://open.spotify.com/playlist/5GhQiRkGuqzpWZSE7OU4Se', #100 songs (unofficial playlist)
    2022: 'https://open.spotify.com/playlist/56r5qRUv3jSxADdmBkhcz7' #100 songs (unofficial playlist)
}
```

```python
# extract the top 100 songs from a Spotify playlist URL

def get_top_100_songs_from_playlist(playlist_url):
    results = sp.playlist_tracks(playlist_url, limit=100)
    top_songs = results['items']
    return top_songs
```

```python
# retrieve the audio features for a given track

def get_audio_features(track_id):
    audio_features = sp.audio_features(track_id)[0]
    return audio_features
```

```python
# List of audio feature names to extract
audio_features_list = [
    'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'duration_ms', 'time_signature']
```

```python
# Loop over the playlist URLs, retrieve the top 100 songs for each playlist, and collect their audio features

data = []

# Loop through the dictionary, retrieve the top 100 songs for each playlist URL, and collect their audio features
for year, playlist_url in playlist_urls.items():
    top_songs = get_top_100_songs_from_playlist(playlist_url)
    for song in top_songs:
        track_name = song['track']['name']
        artist_name = song['track']['artists'][0]['name']
        #other possibilities
        #genre = song['track']['artists'][0]['followers'] *returns an array and artist may not be classified
        #followers = song['track']['artists'][0]['followers']['total']
        track_id = song['track']['id']
        audio_features = get_audio_features(track_id)

        if audio_features:
            # Extract specific audio features and store them as separate columns
            audio_feature_values = {feature: audio_features[feature] for feature in audio_features_list}
            data.append({
                'year': year,
                'track_name': track_name,
                'artist_name': artist_name,
                'track_id': track_id,
                **audio_feature_values
            })

```

```python
# labelling these songs as hits
df = pd.DataFrame(data)
df['hit'] = 1
df
```

```python
df = df.drop_duplicates(subset="track_id", keep="first", inplace=False)
```

```python
spotify_million = spotify_million[(spotify_million.year < 2023) & (spotify_million.year > 2012)]
spotify_million.head()
```

```python
#selecting random songs from the data set without duplicating the hits

import random

# get the list of track IDs from df
other_track_ids = df['track_id'].tolist()

# initialize an empty list to store the randomly selected songs
random_track_ids = []

# group spotify_million by 'year'
grouped_by_year = spotify_million.groupby('year')

# randomly select 500 songs from each year without overlap
num_songs_per_year = 500

for year, group in grouped_by_year:
    available_track_ids = [track_id for track_id in group['track_id'].tolist() if track_id not in other_track_ids]
    selected_track_ids = random.sample(available_track_ids, min(num_songs_per_year, len(available_track_ids)))
    random_track_ids.extend(selected_track_ids)
    other_track_ids.extend(selected_track_ids)



```

```python
spotify_million_sample = spotify_million[spotify_million['track_id'].isin(random_track_ids)].copy()
```

```python
# to check if there are any duplicate songs in the two datasets
check_id = spotify_million_sample['track_id'].tolist()
duplicate_songs = df[df['track_id'].isin(check_id)]
duplicate_songs
```
